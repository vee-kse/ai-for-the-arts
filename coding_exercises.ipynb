## Hello, World! (printing)

I started with the simplest possible program, to check that Codespaces, Jupyter and Copilot all work together.

**Prompt given to Copilot:**

> Could you suggest a python code for 'Hello, world!' to demonstrate the concept of printing to a complete beginner in programming? 

```python
Absolutely! A classic first Python program is:

print("Hello, world!")

How it works

Let's break it down:

print() is a built-in Python function.
The purpose of print() is to display information on the screen.
"Hello, world!" is a piece of text called a string.
The quotation marks (" ") tell Python that the text inside should be treated as characters rather than code.

When you run the program, Python displays:

Hello, world!

Step-by-step explanation
Python reads the line from left to right.
It sees the print() function.
It takes the text inside the parentheses.
It outputs that text to the screen.
Experiment with it

Try changing the text:

print("Hello, Evie!")


Output:

Hello, Evie!


Or print something completely different:

print("I am learning Python!")


Output:

I am learning Python!

Key beginner concepts demonstrated

This simple program introduces:

Functions: print() performs an action.
Strings: Text enclosed in quotation marks.
Syntax: The rules Python follows when writing code.
Program execution: Instructions are carried out line by line.
A slightly larger example
print("Hello, world!")
print("Welcome to Python.")
print("Programming is fun!")


Output:

Hello, world!
Welcome to Python.
Programming is fun!


This demonstrates that Python executes each print() statement one after another.

A good way to think about print() is that it's Python's way of "speaking" to the user by displaying messages on the screen.
```
