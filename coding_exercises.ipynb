# Hello, World! Using a Loop in Python
 
A simple way to introduce **loops** is to repeat the "Hello, world!" message multiple times automatically.
 
## Code
 
```python
for i in range(3):
print("Hello, world!")
```
 
## How It Works
 
Let's break it down:
 
- `for` starts a **loop**.
- `i` is a variable that changes each time the loop runs.
- `range(3)` tells Python to repeat the loop **3 times**.
- `print("Hello, world!")` is the instruction that is repeated.
 
When you run the program, Python displays:
 
```text
Hello, world!
Hello, world!
Hello, world!
```
 
## Step-by-Step Explanation
 
1. Python sees the `for` loop.
2. `range(3)` creates three values: `0`, `1`, and `2`.
3. The loop runs once for each value.
4. Each time the loop runs, `print("Hello, world!")` is executed.
5. The message is displayed three times.
 
## Experiment with It
 
Try changing the number:
 
```python
for i in range(5):
print("Hello, world!")
```
 
### Output
 
```text
Hello, world!
Hello, world!
Hello, world!
Hello, world!
Hello, world!
```
 
Or print a different message:
 
```python
for i in range(3):
print("I am learning Python!")
```
 
### Output
 
```text
I am learning Python!
I am learning Python!
I am learning Python!
```
 
## Key Beginner Concepts Demonstrated
 
- **Loops:** Repeat code automatically.
- **`for` loops:** Repeat a set number of times.
- **`range()`:** Generates a sequence of numbers for the loop.
- **Indentation:** The indented code belongs to the loop and is repeated.
- **Program execution:** Python performs the indented instructions once per loop iteration.
 
## A Slightly Larger Example
 
```python
for i in range(3):
print("Hello, world!")
print("Welcome to Python.")
```
 
### Output
 
```text
Hello, world!
Welcome to Python.
Hello, world!
Welcome to Python.
Hello, world!
Welcome to Python.
```
 
This demonstrates that everything indented inside the loop is repeated together each time the loop runs.
 
> A good way to think about a loop is that it's Python's way of saying, **"Repeat these instructions for me so I don't have to write them multiple times."**

# Hello, World! Using a List in Python

A simple way to introduce **lists** is to store multiple messages in a single variable and then retrieve one of them.

## Code

```python
messages = ["Hello, world!", "Welcome to Python!", "Programming is fun!"]

print(messages[0])
```

## How It Works

Let's break it down:

- `messages` is a variable that stores a **list**.
- A list is a collection of items enclosed in square brackets `[]`.
- Each item in the list is separated by a comma `,`.
- `messages[0]` accesses the first item in the list.
- `print()` displays that item on the screen.

When you run the program, Python displays:

```text
Hello, world!
```

## Step-by-Step Explanation

1. Python creates a list containing three messages.
2. The list is stored in the variable `messages`.
3. `messages[0]` retrieves the first item from the list.
4. `print()` displays that item on the screen.

## Understanding List Positions

Lists use **indexes** (positions) to access items.

```python
messages = ["Hello, world!", "Welcome to Python!", "Programming is fun!"]

print(messages[0])
print(messages[1])
print(messages[2])
```

### Output

```text
Hello, world!
Welcome to Python!
Programming is fun!
```

Notice that the first item has index `0`, not `1`.

## Experiment with It

Try adding another message:

```python

    "Lists are useful!"
]
```

### Output

```text

```

## Key Beginner Concepts Demonstrated

- **Lists:** Store multiple values in one variable.
- **Items (Elements):** Individual values inside a list.
- **Square Brackets `[]`:** Used to create a list.
- **Indexes:** Positions used to access items.
- **Zero-Based Indexing:** The first item is at position `0`.

## A Slightly Larger Example

```python

    print(message)
```

### Output

```text

```

This demonstrates that lists can be used together with loops to process multiple items automatically.

> A good way to think about a list is that it's like a shopping list: one list can hold many items, and each item has a specific position that allows you to find it later.

# Hello, World! Using a Dictionary in Python

A simple way to introduce **dictionaries** is to store information using a **key** and a **value**. This lets you look up data by name rather than by position.

## Code

```python
message = {

}

```

Let's break it down:

- `message` is a variable that stores a **dictionary**.
- Dictionaries use curly braces `{}`.
- `"greeting"` is a **key**.
- `"Hello, world!"` is the **value** associated with that key.
- `message["greeting"]` retrieves the value stored under the key `"greeting"`.
- `print()` displays that value on the screen.

When you run the program, Python displays:

```text
Hello, world!
```

## Step-by-Step Explanation

1. Python creates a dictionary.
2. The dictionary contains one key-value pair:
   - Key: `"greeting"`
   - Value: `"Hello, world!"`
3. The dictionary is stored in the variable `message`.
4. `message["greeting"]` retrieves the value associated with the key `"greeting"`.
5. `print()` displays the value on the screen.

## Understanding Keys and Values

- The **key** is like a word you look up.
- The **value** is the definition you find.

4. The `random` library generates a random number.

Libraries allow programmers to use code written by others instead of creating everything from scratch.

# Hello, World! Using `if`, `elif`, and `else` in Python

A simple way to introduce **conditional statements** is to make the program choose what message to display based on a value.

## Code

```python
hour = 14

if hour < 12:
    print("Good morning!")
elif hour < 18:
    print("Hello, world!")
else:
    print("Good evening!")
```

## How It Works

Let's break it down:

- `hour = 14` stores a value in the variable `hour`.
- `if` checks whether a condition is true.
- `elif` (short for "else if") checks another condition if the previous one was false.
- `else` runs if none of the previous conditions were true.
- `print()` displays the selected message.

When you run the program, Python displays:

```text
Hello, world!
```

This is because `14` is less than `18` but not less than `12`.

## Step-by-Step Explanation

1. Python stores the value `14` in the variable `hour`.
2. It checks the first condition: `hour < 12`.
3. Since `14 < 12` is false, Python moves to the next condition.
4. It checks `hour < 18`.
5. Since `14 < 18` is true, Python executes:

```python
print("Hello, world!")
```

6. The remaining conditions are skipped.

## Key Beginner Concepts Demonstrated

- **Variables:** Store information such as numbers and text.
- **Conditions:** Expressions that evaluate to either `True` or `False`.
- **`if`:** Runs code when a condition is true.
- **`elif`:** Checks additional conditions.
- **`else`:** Provides a default action when no conditions are true.
- **Decision Making:** Programs can choose different actions based on data.

# Hello, World! Using Operators in Python

A simple way to introduce **operators** is to use them to combine and repeat the classic "Hello, world!" message.

## Code

```python
message = "Hello, " + "world!"

print(message)
print(message * 3)
```

## How It Works

Let's break it down:

- `+` is the **addition operator**.
- When used with strings, `+` joins (or **concatenates**) text together.
- `*` is the **multiplication operator**.
- When used with a string and a number, `*` repeats the string multiple times.
- `print()` displays the result on the screen.

When you run the program, Python displays:

```text
Hello, world!
Hello, world!Hello, world!Hello, world!
```

## Step-by-Step Explanation

1. Python evaluates:

   ```python
   "Hello, " + "world!"
   ```

2. The two strings are joined together to create:

   ```text
   Hello, world!
   ```

3. The result is stored in the variable `message`.

4. `print(message)` displays the message.

5. Python evaluates:

   ```python
   message * 3
   ```

6. The string is repeated three times.

7. The repeated string is displayed on the screen.

## Addition with Strings

The `+` operator joins strings together.

```python
greeting = "Hello"
name = "Evie"

print(greeting + " " + name)
```

### Output

```text
Hello Evie
```

This is called **string concatenation**.

## Addition with Numbers

When used with numbers, the `+` operator performs arithmetic addition.

```python
a = 5
b = 3

print(a + b)
```

### Output

```text
8
```

Here, Python adds the two numbers together.

## Multiplication with Numbers

When used with numbers, the `*` operator performs arithmetic multiplication.

```python
print(5 * 3)
```

### Output

```text
15
```

The numbers are multiplied together.

## Multiplication with Strings

When used with a string and a number, the `*` operator repeats the string.

```python
print("Hello! " * 3)
```

### Output

```text
Hello! Hello! Hello!
```

This can be useful when you want the same text to appear multiple times.

## Common Beginner Mistake

Trying to add a string and a number directly causes an error:

```python
print("Hello" + 3)
```

### Result

```text

```

This happens because Python does not know how to combine text and numbers automatically.

To fix it, convert the number to a string:

```python
print("Hello " + str(3))
```

### Output

```text
Hello 3
```

## Key Beginner Concepts Demonstrated

- **Operators:** Symbols that perform actions on values.
- **`+` with numbers:** Adds values together.
- **`+` with strings:** Joins text together.
- **`*` with numbers:** Multiplies values.
- **`*` with strings:** Repeats text.
- **Variables:** Store values for later use.
- **Data Types:** Numbers and strings behave differently with operators.

# Hello, World! Using Classes, Objects, and Methods in Python

A simple way to introduce **classes**, **objects**, and **methods** is to create an object whose job is to display the classic "Hello, world!" message.

## Code

```python
class Greeter:
    def say_hello(self):
        print("Hello, world!")

greeter = Greeter()
greeter.say_hello()
```

## How It Works

Let's break it down:

- `class Greeter:` defines a new **class** called `Greeter`.
- A class acts as a blueprint for creating objects.
- `def say_hello(self):` defines a **method** inside the class.
- A method is a function that belongs to a class.
- `print("Hello, world!")` displays the message.
- `greeter = Greeter()` creates an **object** from the class.
- `greeter.say_hello()` calls the object's method.

When you run the program, Python displays:

```text
Hello, world!
```

## Step-by-Step Explanation

1. Python creates a class called `Greeter`.
2. The class contains a method named `say_hello()`.
3. Python creates an object called `greeter`.
4. The object is based on the `Greeter` class.
5. Python runs `greeter.say_hello()`.
6. The method prints `"Hello, world!"`.

## Understanding Classes and Objects

Think of a class as a blueprint for a house.

```python
class Greeter:
    pass
```

The blueprint describes what can be created.

An object is the actual house built from that blueprint.

```python
greeter = Greeter()
```

You can create many objects from the same class.

```python
greeter1 = Greeter()
greeter2 = Greeter()
```

## Understanding Methods

A method is an action that an object can perform.

```python
class Greeter:
    def say_hello(self):
        print("Hello, world!")
```

Here, `say_hello()` is a method because it belongs to the `Greeter` class.

You call it using:

```python
greeter.say_hello()
```

## What Does `self` Mean?

You may notice this line:


# Hello, World! Through the Lenses of Simplicity, Security, and Inclusion

The classic "Hello, world!" program can be adapted to demonstrate different programming principles. Below are three versions that emphasize **simplicity**, **security**, and **inclusion**, along with explanations of why each approach is valuable.

---

# 1. Simplicity

The simplest code is often the easiest to understand, maintain, and debug.

## Code

```python
print("Hello, world!")
```

## How It Works

- `print()` is a built-in Python function.
- `"Hello, world!"` is a string (text).
- Python displays the text on the screen.

### Output

```text
Hello, world!
```

## Why This Demonstrates Simplicity

- Uses only one statement.
- No unnecessary variables or functions.
- Easy for beginners to read.
- Clearly shows the relationship between code and output.

## Key Principle

> Write the simplest solution that solves the problem.

For a beginner, simple code is usually easier to understand than clever or complex code.

---

# 2. Security

A common security principle is:

> Never trust external input.

The following example safely handles user input while keeping the program simple.

## Code

```python
name = input("Enter your name: ")

print(f"Hello, {name}!")
```

## How It Works

- `input()` asks the user for information.
- The user's response is stored in `name`.
- An f-string inserts the value of `name` into the output.

### Example

```text
Enter your name: Evie
Hello, Evie!
```

## Why This Demonstrates Security

This example treats user input as ordinary text.

For example, if a user enters:

```text
<script>alert("Hello")</script>
```

the program simply displays:

```text
Hello, <script>alert("Hello")</script>!
```

The text is not executed as code.

## Key Security Lessons

- User input should be treated as data, not instructions.
- Avoid writing programs that execute input as Python code.
- Never use functions such as `eval()` on untrusted input.

### Unsafe Example

```python
user_input = input("Enter something: ")
eval(user_input)
```

This allows a user to run arbitrary Python code and should generally be avoided.

## Key Principle

> Accept input carefully and treat it as data unless you have a very good reason not to.

---

# 3. Inclusion

Inclusive software recognizes that users may have different languages, cultures, and backgrounds.

## Code

```python
greetings = {
    "English": "Hello, world!",
    "Spanish": "¡Hola, mundo!",
    "French": "Bonjour le monde !"
}

language = "English"

print(greetings[language])
```

## How It Works

- A dictionary stores greetings in multiple languages.
- The variable `language` selects which greeting to display.
- The selected greeting is printed.

### Output

```text
Hello, world!
```

## Why This Demonstrates Inclusion

- Supports multiple languages.
- Recognizes that users may not all speak English.
- Makes software more accessible to diverse audiences.

### Experiment

Change:

```python

```

to:

```python

```

Output:

```text

```

Or:

```python

```

Output:

```text
Bonjour le monde !
```

## Key Principle

> Design software with different users in mind rather than assuming everyone has the same needs.

# Key Takeaways

- **Simplicity** makes programs easier to read, learn, and maintain.

> A useful rule of thumb is: write code that is simple enough to understand, secure enough to trust, and inclusive enough to welcome all users.

## Algorithms in my everyday life

**Crossing the road**
I stop at the kerb and look right, then left, then right again. If a car is coming, I wait and repeat the check. If the road is clear, I walk across and keep looking as I go. The loop runs until the road is clear, and then the crossing starts. If the crossing has a signal, I press the button first and wait for the green man.

**Cooking dinner**
I pick a recipe and check my ingredients against it. If one is missing, I swap it or buy it. Then I follow the steps in order: prep, cook, season, taste. Some steps depend on others, so the pasta water goes on before I start the sauce. I stop when the food is hot and tastes right.

**Finding a page of my diary**
My diary has dated pages, so I use binary search. I open it near the middle and check the date. If my target date is earlier, I look in the first half. If it is later, I look in the second half. I split the remaining pages in half each time until I land on the right one. This is much faster than flipping through every page from the front.

**Sorting laundry**
I go through the pile one item at a time and put each item in a group by colour and fabric: whites, darks, colours, delicates. Each item gets looked at once. This is a bucket sort. Once every item has a pile, I wash each pile at the right temperature.

**Finding the fastest route between lectures**
I list the possible paths between the two buildings and estimate the time for each. I count walking distance, stairs, and busy corridors. I pick the path with the lowest time. If a road is closed or a lift is out, I drop that path and pick the next best. This works like a shortest path search, where each path has a cost and I choose the cheapest one.
